In the following, we provide a code for the 6th step of the proof with genus g=4.

```python
import numpy as np
import pandas as pd
```

```python
degree=8 #degree of the polynomials Q is 2g
coeff_candidate=[1,0,0,-1,0,0,0,-1,0,0,-1] #candidate polynomial is x^10-x^7-x^3-1
lmin=max(abs(np.roots(coeff_candidate)))
```

We now compute the bounds on the power sums. Since the polynomials we are looking for are skew-reciprocal, we only need to compute half the coefficients and hence half the power sums.

```python
p_up=np.array([],dtype=int)
for i in range(1,int(np.floor(degree/2))+1):
    p_up=np.append(p_up,[int(np.floor(degree/2*(lmin**i+lmin**(-i))))])

print(p_up) #upper bounds on the power sums
```

```python
p_low=np.array([],dtype=int)
for i in range(1,int(np.floor(degree/2))+1):
    p_low=np.append(p_low,[int(np.ceil(min(2-degree,-(degree/2-2)*lmin**i-degree/2*lmin**(-i))))])

print(p_low)
```

We now use the Newton's formula to compute the coefficients and check if the largest root is smaller lmin in absolute value

```python
working=[]  #Create an empty list to push the possibilities at the end
for p1 in range(p_up[0],p_low[0]-1,-1):
    c1=-p1  #Newton formula (idem for other loops)
    p2_first=(-c1*p1)%2   #compute m from lemma 2.2.5 (idem for other loops)
    for p2 in range(p_up[1]-(p_up[1]-p2_first)%2,p_low[1]-1,-2):
        c2=(p1**2-p2)//2
        p3_first=(-c1*p2-c2*p1)%3
        for p3 in range(p_up[2]-(p_up[2]-p3_first)%3,p_low[2]-1,-3):
            c3=(-c1*p2-c2*p1-p3)//3
            p4_first=(-c1*p3-c2*p2-c3*p1)%4
            for p4 in range(p_up[3]-(p_up[3]-p4_first)%4,p_low[3]-1,-4):
                c4=(-c1*p3-c2*p2-c3*p1-p4)//4
                coeff=[1,c1,c2,c3,c4,-c3,c2,-c1,1] #Coefficients of the polynomial using lemma 2.2.4
                roots=np.roots(coeff) #Compute the roots of the polynomial
                l=max(abs(roots))     #Compute the maximal root of the polynomial in absolute value
                if l<lmin+0.00001:    #Check the maximal root is smaller than 1.2044 (Here we work with floats. Hence we introduce a margin to ensure no eliminating a polynomial to numerical errors)
                    coeff.extend([l,p1,p2,p3,p4])  #Push all the information about the polynomial in the list named coeff
                    working.append(coeff)  #Push the list coeff in the list working
```

```python
data=pd.DataFrame(working,columns=['c0','c1','c2','c3','c4','c5','c6','c7','c8','max root','p1','p2','p3','p4'])
print(data)
data.to_excel('candidates_Q.xlsx')
```

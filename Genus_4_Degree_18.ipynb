In the following, we provide a code for the 4th step of the proof with genus g=4, considering polynomials of degree 18 and doing the automated eliminations.

```python
import numpy as np
import pandas as pd
from tqdm import tqdm
from scipy.linalg import companion
from sympy import divisors
from operator import add
```

We first compute the list of polynomials P with degree 18.

```python
degree=18
coeff_candidate=[1,0,0,-1,0,0,0,-1,0,0,-1] #candidate polynomial is x^10-x^7-x^3-1
lmin=max(abs(np.roots(coeff_candidate))) #associated dilatation
```

We now compute the bounds on the power sums. Since the polynomials we are looking for are skew-reciprocal, we only need to compute half the coefficients and hence half the power sums.

```python
p_up=np.array([],dtype=int)
for i in range(1,int(np.floor(degree/2))+1):
    p_up=np.append(p_up,[int(np.floor(degree/2*(lmin**i+lmin**(-i))))])

print(p_up) #upper bounds on the power sums
```

```python
p_low=np.array([],dtype=int)
for i in range(1,int(np.floor(degree/2))+1):
    p_low=np.append(p_low,[int(np.ceil(min(2-degree,-(degree/2-2)*lmin**i-degree/2*lmin**(-i))))])

print(p_low) #lower bounds on the power sums
```

By Lemma 7, we know the traces of the odd power of the homeomorphisms must be non-negativ. Hence, knowing the power sums for Q, we can redifine the lower bounds for P.

```python
p_low[0]=0
p_low[2]=-3
p_low[4]=0
p_low[6]=-7
p_low[8]=-12
```

We now use the Newton's formula to compute the coefficients. We check if the largest root is smaller lmin in absolute value and that it is Perron.

```python
working=[]
for p1 in range(0,2,1): 
    c1=-p1
    p2_first=(-c1*p1)%2
    for p2 in tqdm(range(p_up[1]-(p_up[1]-p2_first)%2,p_low[1]-1,-2)):
        c2=(p1**2-p2)//2
        p3_first=(-c1*p2-c2*p1)%3
        for p3 in range(p_up[2]-(p_up[2]-p3_first)%3,p_low[2]-1,-3):
            c3=(-c1*p2-c2*p1-p3)//3
            p4_first=(-c1*p3-c2*p2-c3*p1)%4
            for p4 in range(p_up[3]-(p_up[3]-p4_first)%4,p_low[3]-1,-4):
                c4=(-c1*p3-c2*p2-c3*p1-p4)//4
                p5_first=(-c1*p4-c2*p3-c3*p2-c4*p1)%5
                for p5 in range(p_up[4]-(p_up[4]-p5_first)%5,p_low[4]-1,-5):
                    c5=(-c1*p4-c2*p3-c3*p2-c4*p1-p5)//5
                    p6_first=(-c1*p5-c2*p4-c3*p3-c4*p2-c5*p1)%6
                    for p6 in range(p_up[5]-(p_up[5]-p6_first)%6,p_low[5]-1,-6):
                        c6=(-c1*p5-c2*p4-c3*p3-c4*p2-c5*p1-p6)//6
                        p7_first=(-c1*p6-c2*p5-c3*p4-c4*p3-c5*p2-c6*p1)%7
                        for p7 in range(p_up[6]-(p_up[6]-p7_first)%7,p_low[6]-1,-7):
                            c7=(-c1*p6-c2*p5-c3*p4-c4*p3-c5*p2-c6*p1-p7)//7
                            p8_first=(-c1*p7-c2*p6-c3*p5-c4*p4-c5*p3-c6*p2-c7*p1)%8
                            for p8 in range(p_up[7]-(p_up[7]-p8_first)%8,p_low[7]-1,-8):
                                c8=(-c1*p7-c2*p6-c3*p5-c4*p4-c5*p3-c6*p2-c7*p1-p8)//8
                                p9_first=(-c1*p8-c2*p7-c3*p6-c4*p5-c5*p4-c6*p3-c7*p2-c8*p1)%9
                                for p9 in range(p_up[8]-(p_up[8]-p9_first)%9,p_low[8]-1,-9):
                                    c9=(-c1*p8-c2*p7-c3*p6-c4*p5-c5*p4-c6*p3-c7*p2-c8*p1-p9)//9
                                    coeff=[1,c1,c2,c3,c4,c5,c6,c7,c8,c9,-c8,c7,-c6,c5,-c4,c3,-c2,c1,-1]
                                    roots=np.roots(coeff)
                                    l=max(roots,key=abs)
                                    l2=sorted(abs(roots))[-2]
                                    if abs(l)<lmin+0.00001 and l.real>1+np.log(2)/(4*degree) and l.imag==0 and np.polyval(coeff,lmin)>-0.00001 and l2<abs(l):
                                        coeff.extend([abs(l),p1,p2,p3,p4,p5,p6,p7,p8,p9])
                                        working.append(coeff)
```

```python
polyP=pd.DataFrame(working,columns=['c0','c1','c2','c3','c4','c5','c6','c7','c8','c9','c10','c11','c12','c13','c14','c15','c16','c17','c18','max root','p1','p2','p3','p4','p5','p6','p7','p8','p9'])
polyP.to_excel('polyP_18.xlsx')
```

```python
polyP=pd.read_excel('polyP_18.xlsx',index_col=0,usecols='A:T,V:AD',dtype=np.int64)
maxrootP=pd.read_excel('polyP_18.xlsx',index_col=0,usecols='A,U')
polyP=polyP.drop([0,1,3,4,9,12,14,16,17,22,32,36,45,46,55,60,63,71,73,76,127,129,134,138,147,160\
                 ,177,178,193,210,252,254,280,303,304,307,310,312,322,323,325,328,334,340,341,344\
                 ,346,357,370,414,437,471,479,482,483,492,493,498,499,506,507,547,553,554,558,562\
                 ,565,566,574,580,584,589,590])
```

We compute a few power sums

```python
maxpowersum=37

zeros=np.zeros(len(polyP.index),dtype=np.int64)
for i in range(10,maxpowersum):
    polyP.insert(len(polyP.columns),f'p{i}',zeros)

for idq in polyP.index:
    c=[polyP[f'c{i}'][idq] for i in range(0,19,1)]
    C=companion(c)
    B=companion(c)
    polyP.at[idq,'p1']=np.matrix.trace(C)
    for i in range(2,maxpowersum):
        B=np.matmul(B,C)
        polyP.at[idq,f'p{i}']=np.matrix.trace(B)
```

We get the polynomials Q we calculated.

```python
polyQ=pd.read_excel('candidates_Q.xlsx',index_col=0,usecols='A:J,L:O',dtype=np.int64)
maxrootQ=pd.read_excel('candidates_Q.xlsx',index_col=0,usecols='A,K')
```

```python
for idq in polyQ.index:  #compute power sums using compagnon matrix 
    c=[polyQ[f'c{i}'][idq] for i in range(0,9,1)]
    C=companion(c)
    C=np.array(C,dtype=np.int64)
    B=companion(c)
    B=np.array(B,dtype=np.int64)
    polyQ.at[idq,'p1']=np.matrix.trace(C)
    for i in range(2,maxpowersum):
        B=np.matmul(B,C)
        polyQ.at[idq,f'p{i}']=np.matrix.trace(B)
```

We now eliminate all combinations PQ for which an odd power sum is negative.

```python
working=[]
for idp in polyP.index:
    for idq in polyQ.index:
        if maxrootQ['max root'][idq]<maxrootP['max root'][idp]+0.00001 and polyP["p1"][idp]+polyQ["p1"][idq]>=0 and polyP["p3"][idp]+polyQ["p3"][idq]>=0\
        and polyP["p5"][idp]+polyQ["p5"][idq]>=0 and polyP["p7"][idp]+polyQ["p7"][idq]>=0 and polyP["p9"][idp]+polyQ["p9"][idq]>=0\
        and polyP["p11"][idp]+polyQ["p11"][idq]>=0 and polyP["p13"][idp]+polyQ["p13"][idq]>=0 and polyP["p15"][idp]+polyQ["p15"][idq]>=0\
        and polyP["p17"][idp]+polyQ["p17"][idq]>=0 and polyP["p19"][idp]+polyQ["p19"][idq]>=0 and polyP["p21"][idp]+polyQ["p21"][idq]>=0\
        and polyP["p23"][idp]+polyQ["p23"][idq]>=0 and polyP["p25"][idp]+polyQ["p25"][idq]>=0 and polyP["p27"][idp]+polyQ["p27"][idq]>=0\
        and polyP["p29"][idp]+polyQ["p29"][idq]>=0 and polyP["p31"][idp]+polyQ["p31"][idq]>=0 and polyP["p33"][idp]+polyQ["p33"][idq]>=0\
        and polyP["p35"][idp]+polyQ["p35"][idq]>=0:
            working.append([idp,idq])  #Push the couples PQ passing the test in the list working
```

We now apply Lemma 9,10 and 16 to count singularities.

```python
working2=[]          
for row in working:  
    singularities=0  
    order=[]         # List for the number of points with order exactly 2i+1, with i the index in the list
    for p in range(1,maxpowersum,2):  # Application of Lemma 16
        d_list=divisors(p)            
        points=polyP[f'p{p}'][row[0]]+polyQ[f'p{p}'][row[1]]  
        for d in range(len(d_list)-1):
            points-=order[(d_list[d]-1)//2] 
        order.append(points)   
        if (points/(p))%2==1:  
            singularities+=p     
    for n in [4,6,8,16]:   # Application of Lemma 9 considering f^4,f^6,f^8,f^16
        delta=polyP[f'p{n}'][row[0]]+polyQ[f'p{n}'][row[1]]-polyP[f'p{2*n}'][row[0]]-polyQ[f'p{2*n}'][row[1]]  
        if delta>0:
            singularities+=delta  
    if singularities<=12:  #If there are less than 12 singularities, we push the polynomial in the working list
        working2.append(row)
working=working2  
```

We do the same test using other powers of f for the Lemma 9.

```python
working2=[]
for row in working:
    singularities=0
    order=[] 
    for p in range(1,maxpowersum,2):
        d_list=divisors(p)
        points=polyP[f'p{p}'][row[0]]+polyQ[f'p{p}'][row[1]]
        for d in range(len(d_list)-1):
            points-=order[(d_list[d]-1)//2]
        order.append(points)
        if (points/(p))%2==1:
            singularities+=p
    for n in [2,4,8,16]:
        delta=polyP[f'p{n}'][row[0]]+polyQ[f'p{n}'][row[1]]-polyP[f'p{2*n}'][row[0]]-polyQ[f'p{2*n}'][row[1]]        
        if delta>0:
            singularities+=delta
    if singularities<=12:
        working2.append(row)
working=working2
```

We now use Lemma 10 instead of Lemma 9.

```python
deleted=[] 
for row in working:
    singularities=0
    order=[] 
    for p in range(1,maxpowersum,2):
        d_list=divisors(p)
        points=polyP[f'p{p}'][row[0]]+polyQ[f'p{p}'][row[1]]
        for d in range(len(d_list)-1):
            points-=order[(d_list[d]-1)//2]
        order.append(points)
        if (points/(p))%2==1:
            singularities+=p
    for n in [1,3,5,7,9,11,13]: #Application of Lemma 10
        if order[(n-1)//2]==0:  
            delta= 2-polyP[f'p{2*n}'][row[0]]-polyQ[f'p{2*n}'][row[1]]  
        elif order[(n-1)//2]%2==1: 
            delta= 2-polyP[f'p{2*n}'][row[0]]-polyQ[f'p{2*n}'][row[1]]+polyP[f'p{n}'][row[0]]+polyQ[f'p{n}'][row[1]]+4*n 
        else:
            break
        if not(delta<0 or (delta>=0 and singularities+delta<=12)): 
            deleted.append(row)
working = [x for x in working if x not in deleted]
```

```python
eliminated=[]
for row in working:
    for p in range(1,maxpowersum):
        if polyP[f'p{p}'][row[0]]+polyQ[f'p{p}'][row[1]]<-10:
            eliminated.append(row)
            break
working=[x for x in working if x not in eliminated]
```

```python
for p in range(2,maxpowersum//3,2):
    eliminated=[]
    for row in working:
        if polyP[f'p{p}'][row[0]]+polyQ[f'p{p}'][row[1]]==1 and polyP[f'p{3*p}'][row[0]]+polyQ[f'p{3*p}'][row[1]]<-4\
        or polyP[f'p{p}'][row[0]]+polyQ[f'p{p}'][row[1]]==0 and polyP[f'p{3*p}'][row[0]]+polyQ[f'p{3*p}'][row[1]]<2\
        or polyP[f'p{p}'][row[0]]+polyQ[f'p{p}'][row[1]]==-1 and polyP[f'p{3*p}'][row[0]]+polyQ[f'p{3*p}'][row[1]]<8\
        or polyP[f'p{p}'][row[0]]+polyQ[f'p{p}'][row[1]]==-2 and polyP[f'p{3*p}'][row[0]]+polyQ[f'p{3*p}'][row[1]]<14\
        or polyP[f'p{p}'][row[0]]+polyQ[f'p{p}'][row[1]]==-3 and polyP[f'p{3*p}'][row[0]]+polyQ[f'p{3*p}'][row[1]]<20\
        or polyP[f'p{p}'][row[0]]+polyQ[f'p{p}'][row[1]]==-4 and polyP[f'p{3*p}'][row[0]]+polyQ[f'p{3*p}'][row[1]]<26\
        or polyP[f'p{p}'][row[0]]+polyQ[f'p{p}'][row[1]]==-5 and polyP[f'p{3*p}'][row[0]]+polyQ[f'p{3*p}'][row[1]]<32\
        or polyP[f'p{p}'][row[0]]+polyQ[f'p{p}'][row[1]]==-6 and polyP[f'p{3*p}'][row[0]]+polyQ[f'p{3*p}'][row[1]]<38\
        or polyP[f'p{p}'][row[0]]+polyQ[f'p{p}'][row[1]]==-7 and polyP[f'p{3*p}'][row[0]]+polyQ[f'p{3*p}'][row[1]]<44\
        or polyP[f'p{p}'][row[0]]+polyQ[f'p{p}'][row[1]]==-8 and polyP[f'p{3*p}'][row[0]]+polyQ[f'p{3*p}'][row[1]]<50\
        or polyP[f'p{p}'][row[0]]+polyQ[f'p{p}'][row[1]]==-9 and polyP[f'p{3*p}'][row[0]]+polyQ[f'p{3*p}'][row[1]]<56\
        or polyP[f'p{p}'][row[0]]+polyQ[f'p{p}'][row[1]]==-10 and polyP[f'p{3*p}'][row[0]]+polyQ[f'p{3*p}'][row[1]]<62:  #Elimination of the polynomials not fullfilling Lemma
            eliminated.append(row)
    working=[x for x in working if x not in eliminated]
```

```python
table=[]
for row in working:
    power_sum_P=[polyP[f'p{i}'][row[0]] for i in range(1,maxpowersum)]
    power_sum_Q=[polyQ[f'p{i}'][row[1]] for i in range(1,maxpowersum)]
    sum=list(map(add,power_sum_P,power_sum_Q))
    sum.extend(row)
    table.append(sum)
names=[f'p{i}' for i in range(1,maxpowersum)]
names.extend(["idp","idq"])
data=pd.DataFrame(table,columns=names)
data=data.drop_duplicates(subset=names[:-2])#,inplace=True) #ligne à enlever quand on fait l'élimination des cyclotomiques*non primitifs
data.to_excel('powersumsall.xlsx')
```

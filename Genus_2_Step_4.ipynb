In the following, we provide a code for the 4th step of the proof with genus g=2.

```python
import numpy as np
import pandas as pd
```

```python
degree=6 #degree of the polynomials P is at most 6g-6=6
coeff_candidate=[1,-1,0,0,0,-1,-1] #candidate polynomial is x^6-x^5-x-1
lmin=max(abs(np.roots(coeff_candidate))) #associated dilatation
```

We now compute the bounds on the power sums. Since the polynomials we are looking for are skew-reciprocal, we only need to compute half the coefficients and hence half the power sums.

```python
p_up=np.array([],dtype=int)
for i in range(1,int(np.floor(degree/2))+1):
    p_up=np.append(p_up,[int(np.floor(degree/2*(lmin**i+lmin**(-i))))])

print(p_up) #upper bounds on the power sums
```

```python
p_low=np.array([],dtype=int)
for i in range(1,int(np.floor(degree/2))+1):
    p_low=np.append(p_low,[int(np.ceil(min(2-degree,-(degree/2-2)*lmin**i-degree/2*lmin**(-i))))])

print(p_low) #lower bounds
```

By Lemma 7, we know the traces of the odd powers of the pseudo-Anosov homeomorphisms must be non-negative. Since the first and third power sums of the polynomials Q are, respectively, at most 1 and 2, we can redifine the lower bounds here.

```python
p_low[0]=-1
p_low[2]=-2
```

We now use the Newton's formula to compute the coefficients. We check if the largest root is smaller lmin in absolute value and that it is Perron.

```python
working=[]
for p1 in range(p_up[0],p_low[0]-1,-1):
    c1=-p1
    p2_first=(-c1*p1)%2
    for p2 in range(p_up[1]-(p_up[1]-p2_first)%2,p_low[1]-1,-2):
        c2=(p1**2-p2)//2
        p3_first=(-c1*p2-c2*p1)%3
        for p3 in range(p_up[2]-(p_up[2]-p3_first)%3,p_low[2]-1,-3):
            c3=(-c1*p2-c2*p1-p3)//3
            coeff=[1,c1,c2,c3,-c2,c1,-1] 
            roots=np.roots(coeff)
            l=max(roots,key=abs)
            l2=sorted(abs(roots))[-2]
            if abs(l)<lmin+0.00001 and l.real>0 and l.imag==0 and np.polyval(coeff,lmin)>=-0.00001 and l2<abs(l):
                coeff.extend([l,p1,p2,p3])
                working.append(coeff)
```

```python
data=pd.DataFrame(working,columns=['c0','c1','c2','c3','c4','c5','c6','max root','p1','p2','p3'])
print(data)
```

The polynomials 1,4,5 and 6 can all be eliminated since they cleared the tests because of the margins we introduced.

In fact, the first polynomial is the candidate polynomial.
The polynomials 6 is cyclotomic and the polynomials 4 and 5 are non-primitive.

We hence get the following list:

```python
data=data.drop([1,4,5,6])
print(data)
```
